## Teste de performance dos Algoritmos de SSSP

```python
import numpy as np
import pandas as pd
import seaborn as sns

from matplotlib import pyplot as plt
```

#### 1. Configurando parâmetros para rotina de testes

```python
from graphs.graphs import Graph, dijkstra, generate_random_graph
from graphs.duan_et_al import sssp_duan_et_al

# Quantidade de grafos por configuração
num_graphs = 5

# Quantidade de vértices
num_vertices = (100, 1000, 10000)

# Densidade de arestas
edge_probabilities = (0.001, 0.0001, 0.00001)

# Algoritmos
algorithms = (dijkstra, sssp_duan_et_al)
```

#### 2. Executando rotina de testes

```python
from random import seed
from time import perf_counter
import csv

# Garantir reprodutibilidade
seed(42)

results = []

qt_executions = (
    len(num_vertices)
    * len(edge_probabilities)
    * len(algorithms)
    * num_graphs
)

current_execution = 0

for num_v in num_vertices:
    for edge_prob in edge_probabilities:
        for alg in algorithms:
            for _ in range(num_graphs):

                g = generate_random_graph(num_v, edge_prob)

                current_execution += 1
                progress = (current_execution / qt_executions) * 100

                print(
                    f"\rExecuted {progress:.2f}% "
                    f"| num_vertices: {num_v} "
                    f"| edge_probability: {edge_prob} "
                    f"| algorithm: {alg.__name__}",
                    end="",
                    flush=True
                )

                start_time = perf_counter()
                r = alg(g, 0)
                end_time = perf_counter()

                delta_time = end_time - start_time

                results.append({
                    "num_vertices": num_v,
                    "edge_probability": edge_prob,
                    "algorithm": alg.__name__,
                    "execution_time": delta_time
                })

print("\nExecution finished.")

# Salvar os resultados em CSV
with open('data/first_results.csv', 'w', newline='') as f:
    writer = csv.DictWriter(
        f,
        fieldnames=[
            "num_vertices",
            "edge_probability",
            "algorithm",
            "execution_time"
        ]
    )

    writer.writeheader()
    writer.writerows(results)
```
